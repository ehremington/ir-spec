```python
def ExportRefractedAngleData(
    snells, wave_numbers, 
    n0_array, k0_array, 
    n_arrays, k_arrays, 
    th0=45,
    filename='refracted_angle_export.csv'):

    

    # Store results in a dictionary
    data_dict = {'wavenumber': wave_numbers}

    for i, (n_target, k_target) in enumerate(zip(n_arrays, k_arrays)):
        theta_r = snells(n0_array, n_target, k0_array, k_target, th0)
        label = f"thetaR_{i}"
        data_dict[label] = np.real(theta_r)

    df_export = pd.DataFrame(data_dict)
    df_export.to_csv(filename, index=False)
    print(f"Exported to {filename}") 

n_arrays = [
    n0p50_array, n0p55_array, n0p60_array, n0p65_array, n0p70_array,
    n0p75_array, n0p80_array, n0p85_array, n0p90_array, n0p95_array,
    n1p00_array, n1p05_array, n1p10_array, n1p15_array, n1p20_array,
    n1p25_array, n1p30_array, n1p35_array, n1p40_array, n1p45_array,
    n1p50_array, n1p55_array, n1p60_array, n1p65_array, n1p70_array,
    n1p75_array, n1p80_array, n1p85_array, n1p90_array, n1p95_array,
    n2p00_array
]

k_arrays = [
    k0p50_array, k0p55_array, k0p60_array, k0p65_array, k0p70_array,
    k0p75_array, k0p80_array, k0p85_array, k0p90_array, k0p95_array,
    k1p00_array, k1p05_array, k1p10_array, k1p15_array, k1p20_array,
    k1p25_array, k1p30_array, k1p35_array, k1p40_array, k1p45_array,
    k1p50_array, k1p55_array, k1p60_array, k1p65_array, k1p70_array,
    k1p75_array, k1p80_array, k1p85_array, k1p90_array, k1p95_array,
    k2p00_array
]

ExportRefractedAngleData(
    snells, wave_numbers, 
    n0_array, k0_array, 
    n_arrays, k_arrays, 
    th0=80,  # You can change this
    filename="refracted_angles_80deg.csv"
)
```
