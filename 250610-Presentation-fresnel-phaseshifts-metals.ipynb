Basics of our code. 

Speed of light in vaccum $c = 299 * 10^8$

Speed of light in vaccum $v=$material dependant


index of refraction (not complex): the ratio of the speed of light in vaccum to the speed of light in a specific medium: $n=\frac{c}{v}$ 

Snells law: https://www.youtube.com/watch?v=Iq1a_KJTWJ8 : $$n_i sin\theta_i=n_t sin\theta_t$$
snells law also gives us an alternative way to write our index of refraction:$$n_i=\frac{n_t sin\theta_t}{sin\theta_i}$$ 

Imaginary number: $j=\sqrt{-1}$ 

Complex number: a number with both a real and imaginary part: $\tilde{Z}=A+jB$ 

Complex conjugate:$\tilde{Z^*}=a-jb$ 

Multiplying by complex conjugate $\tilde{Z}\tilde{Z^*}=a^2+b^2$ 

Magnitude of complex number: $|\tilde{Z}| =\sqrt{\tilde{Z} \tilde{Z^*}}=\sqrt{a^2+b^2}$ 

Phase shift of a complex number: $\theta=tan^{-1}(\frac{b}{a})$

Absorbtion coefficient: denoted as k. 

index of refraction (complex): $\tilde{n}=n+jk$ 







```python

def snells(ni, nt, ki, kt, thi):
    return(arcsin((ni+1j*ki)/(nt+1j*kt)*sin(thi)))
```

Snells law with complex numbers returning $\theta_t$
$$\tilde{n_i}sin\theta_i=\tilde{n_t}sin\theta_t$$
$$(n_i+jk_1)sin\theta_i=(n_t+jk_t)sin\theta_t$$
$$\frac{(n_i+jk_1)}{(n_t+jk_t)}sin\theta_i=sin\theta_t$$
$$sin^{-1} (\frac{(n_i+jk_1)}{(n_t+jk_t)}sin\theta_i)=\theta_t$$

```python
def costht(ni, nt, ki, kt, thi):
    return(np.sqrt(1-((ni+1j*ki)/(nt+1j*kt)*sin(thi))**2))
    # return(1j*np.sqrt(((ni+1j*ki)/(nt+1j*kt)*sin(thi))**2-1-0j))
```

This is another way for us to calculate and work with $\theta_t$. 
$$sin^{-1} (\frac{(n_i+jk_1)}{(n_t+jk_t)}sin\theta_i)=\theta_t$$ becomes invalid when $n_i>n_t$ because the argument of the inverse sine becomes greater than 1 (total internal reflection. Instead we compute our fresnell coefficients using this substitution: 

$$cos\theta_t=\sqrt{1-(\frac {n_i+ik_i}{n_t+ik_t})^2(sin\theta_i)^2}$$

```python
def rperp(ni, nt, ki, kt, thi, mui=8.85*10**-12, mut=8.85*10**-12):
    '''
    this will provide the reflection coefficient for light 
    perpendicular to the plane of incidence and therefore
    parallel to the interface of the material. 

    this is S-polarized in the nomenclature of optics
    '''
    cos_thi = cos(thi)
    cos_tht = costht(ni, nt, ki, kt, thi)
    return(((ni+1j*ki)/mui*cos_thi-(nt+1j*kt)/mut*cos_tht)/((ni+1j*ki)/mui*cos_thi+(nt+1j*kt)/mut*cos_tht))
```

```python
def Rpara(ni, nt, ki, kt, thi):
    '''
    reflectance for light parallel to the 
    plane of incidence.

    this is P-polarized light
    '''
    # return(rpara(ni, nt, ki, kt, thi)**2)
    r = rpara(ni, nt, ki, kt, thi)
    return(r*r.conjugate())
```
